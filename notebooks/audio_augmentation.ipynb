## Transformations

Loading the input audio file

```python
from src.scripts.audio_augmentation import (change_pitch, change_speed, 
                                            add_reverb, add_echo, change_volume)
import notebooks.help_functions as hf
import soundfile as sf
from src.pipeline.audio_data import AudioData


audio_signal, sample_rate = sf.read(input_wav_path)
audio_data_orginal = AudioData(audio_signal, sample_rate)
hf.display_audio(audio_data_orginal)
```

Changing pitch of audio

```python
audio_data = change_pitch(audio_data_orginal, 5)
hf.display_audio(audio_data)
```

Changing speed of audio

```python
speed = 1.5
audio_data = change_speed(audio_data_orginal, speed)
hf.display_audio(audio_data)
```

Adding reverb

```python
audio_data = add_reverb(audio_data_orginal, 0.2)
hf.display_audio(audio_data)
```

Adding echo

```python
audio_data = add_echo(audio_data_orginal, 0.2, 1)
hf.display_audio(audio_data)
```
